# 10.1 DataFrame kennenlernen

Bisher haben wir Daten in Listen und NumPy-Arrays gespeichert. Listen eignen
sich gut für kleine Sammlungen gemischter Objekte, Arrays für homogene
Zahlenreihen. In der Praxis liegen Daten jedoch häufig als Tabellen vor: Jede
Zeile entspricht einem Datensatz, jede Spalte einer Eigenschaft. Solche Tabellen
enthalten oft sowohl Zahlen als auch Textspalten, und sie können Tausende von
Zeilen umfassen. Listen und Arrays stoßen hier schnell an ihre Grenzen. Das
Modul **Pandas** stellt mit dem **DataFrame** genau den richtigen Datentyp für
diese Aufgabe bereit.

## CSV-Daten einlesen mit `pd.read_csv()`

Pandas wird üblicherweise mit dem Alias `pd` importiert:

```
import pandas as pd
```

Der gebräuchlichste Weg, Daten in Pandas zu laden, ist das Einlesen einer
CSV-Datei. CSV steht für *Comma-Separated Values* und bezeichnet ein einfaches
Textformat, bei dem die Werte jeder Zeile durch ein Trennzeichen voneinander
getrennt sind. Die Funktion `pd.read_csv()` liest eine solche Datei und gibt
einen **DataFrame** zurück, also eine zweidimensionale Tabelle mit Zeilen und
Spalten.

Wir arbeiten in diesem und den folgenden Kapiteln mit einem Datensatz von
AutoScout24, der Fahrzeuginserate aus Deutschland aus dem Jahr 2020 enthält:

```
df = pd.read_csv("autoscout24_DE_2020_cleaned.csv")
```

Die Variable `df` ist die übliche Abkürzung für DataFrame. Der Datensatz ist nun
im Arbeitsspeicher verfügbar und steht für alle weiteren Operationen zur
Verfügung (sofern die Datei nicht extrem groß wird).

**Hinweis: Trennzeichen und Dezimalzeichen**
`pd.read_csv()` erwartet standardmäßig ein Komma als Trennzeichen. Viele
deutschsprachige Programme exportieren CSV-Dateien jedoch mit Semikolon als
Trennzeichen, weil das Komma im Deutschen als Dezimalzeichen verwendet wird. In
solchen Fällen lautet der Aufruf:

`pd.read_csv("datei.csv", sep=";", decimal=",")`

Wenn eine CSV-Datei beim Einlesen nur eine einzige Spalte ergibt, ist ein
falsches Trennzeichen die häufigste Ursache.

## Den DataFrame erkunden

Nach dem Einlesen ist der erste Schritt immer, den DataFrame zu erkunden. Pandas
stellt dafür mehrere Methoden und Attribute bereit.

Mit `.head()` zeigen wir die ersten fünf Zeilen:

```
df.head()
```

In einem Jupyter Notebook wird der DataFrame dabei als formatierte Tabelle
angezeigt, was deutlich übersichtlicher ist als `print()`. Möchte man eine
andere Anzahl von Zeilen sehen, übergibt man die gewünschte Zahl als Argument:
`df.head(10)`.

Das Attribut `.shape` gibt die Abmessungen des DataFrames als Tupel zurück,
also `(Anzahl Zeilen, Anzahl Spalten)`:

```
print(df.shape)
```

Mit `.dtypes` sehen wir den Datentyp jeder Spalte:

```
print(df.dtypes)
```

Die Ausgabe zeigt, dass numerische Spalten wie `Preis (Euro)` oder `Leistung
(kW)` als `int64` gespeichert sind, Textspalten wie `Marke` oder `Kraftstoff`
hingegen als `object`. In unserem Datensatz sind das überwiegend Texte
(Python-Strings).

Die Methode `.describe()` berechnet deskriptive Statistiken für alle numerischen
Spalten auf einmal:

```
df.describe()
```

Die Ausgabe enthält für jede numerische Spalte die Anzahl der Werte (`count`),
den Mittelwert (`mean`), die Standardabweichung (`std`), das Minimum (`min`),
die drei Quartile und das Maximum (`max`). Ein schneller Blick auf `.describe()`
zeigt sofort, ob Werte plausibel sind oder ob Ausreißer vorhanden sind.

## Spalten auswählen und neue Spalten berechnen

Eine einzelne Spalte wählen wir aus, indem wir den Spaltennamen in eckigen
Klammern übergeben:

```
preise = df["Preis (Euro)"]
print(preise.head())
```

Das Ergebnis ist eine **Series**, der eindimensionale Pandas-Datentyp. Eine
Series verhält sich ähnlich wie ein NumPy-Array: Sie unterstützt dieselben
Vektoroperationen und statistischen Methoden.

```
print(preise.mean())   # Mittelwert
print(preise.min())    # Minimum
print(preise.max())    # Maximum
```

Der Vorteil gegenüber einem reinen NumPy-Array: Jede Series trägt ihren
Spaltennamen mit sich, sodass Ergebnisse immer eindeutig zugeordnet werden
können.

Eine neue Spalte fügen wir dem DataFrame hinzu, indem wir einen neuen
Spaltennamen in eckigen Klammern auf der linken Seite einer Zuweisung angeben
und den Wert durch eine Vektoroperation berechnen. Wir berechnen den Preis pro
Kilowatt als Maß dafür, wie teuer die Motorleistung eines Fahrzeugs ist:

```
df["Preis_pro_kW"] = df["Preis (Euro)"] / df["Leistung (kW)"]
df["Preis_pro_kW"].head()
```

Die Vektoroperation wird elementweise auf alle 18566 Zeilen angewendet, ohne
dass eine Schleife nötig ist. Das ist dasselbe Prinzip wie bei NumPy-Arrays aus
Kapitel 9.1. Der neue Spaltenwert ist nun fest im DataFrame gespeichert und
steht für alle weiteren Auswertungen zur Verfügung.
